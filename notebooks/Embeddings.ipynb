```python
import numpy as np
import json
import os

import openai

# To use free, local open-source embeddings, we use the SentenceTransformers library
from sentence_transformers import SentenceTransformer
```

```python
# Retrieve the API key from environment variables
api_key = os.getenv("HF_TOKEN")

# Set the OpenAI API key
openai.api_key = api_key
```

```python
# Load the dataset
with open('../data/ocpq_dataset.json', 'r') as f:
    dataset = json.load(f)

print("Loading the model directly from our local downloaded directory...")
# Load the pre-trained model from the local folder where we just manually downloaded it
model = SentenceTransformer('./local-all-MiniLM-L6-v2') 
print("Model loaded successfully!")
```

```python
def get_embedding(text):
    # sentence-transformers encodes text directly into a numpy array
    embedding = model.encode(text)
    return embedding

# For a small prototype, let's embed all dataset questions:
# extract just the natural language questions
nl_queries = [item['nl'] for item in dataset]

print(f"Encoding {len(nl_queries)} examples. Computing embeddings...")
# encode can process a whole list in a batch very efficiently
# Added show_progress_bar=True so you can track the speed
embeddings = model.encode(nl_queries, show_progress_bar=True)
print("Finished compute.")
```

```python
# Save embeddings to a file using numpy
np.save('../data/embeddings.npy', np.array(embeddings))

# Alternatively, to save as a JSON file:
# with open('embeddings.json', 'w') as f:
#     json.dump(embeddings, f)
```
