```python
import numpy as np
import json
import os

import openai
from openai import OpenAI
```

```python
# Retrieve the API key from environment variables
api_key = os.getenv("OPENAI_API_KEY")

# Set the OpenAI API key
openai.api_key = api_key
```

```python
# Load the dataset
with open('ocpq_dataset.json', 'r') as f:
    dataset = json.load(f)

# Initialize the OpenAI client
client = OpenAI(api_key=api_key)

# Optional: Generate embeddings for all natural language questions in the dataset
# You can save this to a file so you don't have to re-embed every time.
def get_embedding(text):
    response = client.embeddings.create(
        input=text,
        model="text-embedding-ada-002"
    )
    return response.data[0].embedding

# For a small prototype, let's embed all dataset questions:
embeddings = [get_embedding(item['nl']) for item in dataset]
```

```python
# Save embeddings to a file using numpy
np.save('embeddings.npy', np.array(embeddings))

# Alternatively, to save as a JSON file:
# with open('embeddings.json', 'w') as f:
#     json.dump(embeddings, f)
```
